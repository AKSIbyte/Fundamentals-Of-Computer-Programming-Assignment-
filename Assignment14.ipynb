```python
# Q1 Create class for the primary inventories in CSE lab. Also create suitable member variables and member functions for the classes created by you. Create suitable objects of these classes.

class Computer:
    def __init__(self, name, ram, storage):
        self.name = name
        self.ram = ram
        self.storage = storage

    def display_specs(self):
        print(f"Name: {self.name}, RAM: {self.ram}, Storage: {self.storage}")


class Printer:
    def __init__(self, model, type, connection):
        self.model = model
        self.type = type  # e.g., laser, inkjet
        self.connection = connection  # e.g., USB, network

    def print_info(self):
        print(f"Model: {self.model}, Type: {self.type}, Connection: {self.connection}")


class Projector:
    def __init__(self, brand, resolution, brightness):
        self.brand = brand
        self.resolution = resolution
        self.brightness = brightness

    def display_projector_info(self):
        print(f"Brand: {self.brand}, Resolution: {self.resolution}, Brightness: {self.brightness}")


# Creating objects
computer1 = Computer("Dell XPS", "16GB", "512GB SSD")
printer1 = Printer("HP LaserJet", "Laser", "USB")
projector1 = Projector("Epson", "1080p", "3000 lumens")


computer1.display_specs()
printer1.print_info()
projector1.display_projector_info()
```

```python
# Q2 Write code to implement the following. You can make necessary assumptions.
# Create a class Vehicle having the following member variables:
# vehicle_no: vehicle plate number
# color: color of vehicle
# Create an initialization function to initialize the object.
# Create a class Car which inherits all the properties of Vehicle class. Additionally, create the following member variables for Car class:
# engine_type: type of engine
# Create an initialization function to initialize the object.
# Create a class Truck which inherits all the properties of Car class. Additionally, create the following member variable for Truck class:
# load: maximum load that the truck can bear
# Create an initialization function to initialize the object.
# Create a method display() for class Truck to display the vehicle_no, color, engine_type and load for the obj object.
# Create a object obj of Truck class and call the display() method.
# Mention how function overriding is taking place in the code written by you.
# In the code which is the base class/parent class and which are the derived/child classes.

class Vehicle:
    def __init__(self, vehicle_no, color):
        self.vehicle_no = vehicle_no
        self.color = color


class Car(Vehicle):
    def __init__(self, vehicle_no, color, engine_type):
        super().__init__(vehicle_no, color)
        self.engine_type = engine_type


class Truck(Car):
    def __init__(self, vehicle_no, color, engine_type, load):
        super().__init__(vehicle_no, color, engine_type)
        self.load = load

    def display(self):
        print(f"Vehicle Number: {self.vehicle_no}")
        print(f"Color: {self.color}")
        print(f"Engine Type: {self.engine_type}")
        print(f"Maximum Load: {self.load}")


# Create a Truck object
obj = Truck("MH-01-2345", "Red", "Diesel", "10 tons")

# Call the display method
obj.display()
```
