```python
# 4.	Write a program to add two matrices.

import numpy as np

def add_matrices(matrix1, matrix2):
    """Adds two matrices if they have compatible dimensions."""
    try:
        result = np.add(matrix1, matrix2)
        return result
    except ValueError:
        return "Matrices cannot be added. Incompatible dimensions."


# Example usage:
matrix_a = np.array([[1, 2], [3, 4]])
matrix_b = np.array([[5, 6], [7, 8]])

sum_matrix = add_matrices(matrix_a, matrix_b)

if isinstance(sum_matrix, str):
    print(sum_matrix)
else:
    print("Sum of matrices:\n", sum_matrix)


matrix_c = np.array([[1,2,3],[4,5,6]])
matrix_d = np.array([[7,8],[9,10]])

sum_matrix = add_matrices(matrix_c, matrix_d)
if isinstance(sum_matrix, str):
    print(sum_matrix)
else:
    print("Sum of matrices:\n", sum_matrix)
```
