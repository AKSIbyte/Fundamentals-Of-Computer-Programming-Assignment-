```python
# Q1 Questions from dictionary,
# (a)	Create an empty dictionary “D”.
# (b)	Add 3 items in this dictionary “D”, where key is the enrollment number of the student and value is the name of
#               the student.
# (c)	Remove a student details having a specific enrollment number.
# (d)	Create a copy of this dictionary “D1” using the copy() method and display this newly created dictionary.
# (e)	Clear the contents of dictionary D1.
# (f)	Display all the keys of the dictionary “D” using keys() method.
# (g)	Display all the values of the dictionary “D” using values() method.
# (h)	Display all the key, value pair in dictionary “D” using  items() method.
# (i)	Use fromkeys() method to create a new dictionary with keys from a given sequence and values set to a specified value.

# Q1
# (a)
D = {}

# (b)
D[202101] = "Alice"
D[202102] = "Bob"
D[202103] = "Charlie"

# (c)
enrollment_to_remove = 202102
if enrollment_to_remove in D:
  del D[enrollment_to_remove]
  print(f"Student with enrollment {enrollment_to_remove} removed.")
else:
  print(f"Student with enrollment {enrollment_to_remove} not found.")

# (d)
D1 = D.copy()
print("D1:", D1)

# (e)
D1.clear()
print("D1 after clearing:", D1)

# (f)
print("Keys of D:", D.keys())

# (g)
print("Values of D:", D.values())

# (h)
print("Items of D:", D.items())

# (i)
keys_sequence = (1, 2, 3)
specified_value = "Default"
new_dict = dict.fromkeys(keys_sequence, specified_value)
print("New dictionary using fromkeys():", new_dict)


# Q2
NU = {"A1": {}, "A2": {}}

while True:
  name = input("Enter student name (or type 'done' to finish): ")
  if name.lower() == 'done':
    break
  section = input("Enter section (A1 or A2): ")
  enrollment = int(input("Enter enrollment number: "))

  if section.upper() == "A1":
    NU["A1"][enrollment] = {"name": name, "section": section}
  elif section.upper() == "A2":
    NU["A2"][enrollment] = {"name": name, "section": section}
  else:
    print("Invalid section. Please enter A1 or A2.")

NU
```

```python
# Q2. Create a dictionary “NU”  containing two sub dictionaries “A1” and “A2”. Take then name, section and enrollment number as input from user. If the student is from A1 section then insert the details in A1 dictionary else insert in A2 dictionary.

NU = {"A1": {}, "A2": {}}

while True:
  name = input("Enter student name (or type 'done' to finish): ")
  if name.lower() == 'done':
    break
  section = input("Enter section (A1 or A2): ")

  # Get enrollment input.
  while True:
    enrollment = input("Enter enrollment number: ")
    # Check if enrollment is numeric.
    if enrollment.isnumeric():
      enrollment = int(enrollment) # Convert to integer if numeric
      break
    else:
      print("Invalid enrollment number. Please enter a numeric value.")

  if section.upper() == "A1":
    NU["A1"][enrollment] = {"name": name, "section": section}
  elif section.upper() == "A2":
    NU["A2"][enrollment] = {"name": name, "section": section}
  else:
    print("Invalid section. Please enter A1 or A2.")

NU
```
