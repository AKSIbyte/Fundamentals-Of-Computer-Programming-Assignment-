```python
# prompt: Q1 Create a file “abc.txt” in a specific drive or folder. Write the following content in the file:
# I am great.
# I am intelligent.
# I love my family.

# Q1
file_path = "/content/abc.txt"  # Replace with your desired path

with open(file_path, "w") as file:
    file.write("I am great.\n")
    file.write("I am intelligent.\n")
    file.write("I love my family.\n")

# Q2
vowels = "aeiouAEIOU"
vowel_count = 0
dot_count = 0

with open(file_path, "r") as file:
    for line in file:
        for char in line:
            if char in vowels:
                vowel_count += 1
            elif char == ".":
                dot_count += 1

print(f"Number of vowels: {vowel_count}")
print(f"Number of dots: {dot_count}")


# Q3
with open(file_path, "a") as file:
    file.write("I am happy.\n")


# Q4
# w+ mode: Opens a file for both reading and writing. Overwrites the existing file if it exists.
# r+ mode: Opens a file for both reading and writing. Does not overwrite the existing file.

#Demonstrates w+
with open("/content/wplus.txt", "w+") as file:
    file.write("This is w+ mode demonstration\n") #writes to file
    file.seek(0) #sets file pointer to the beginning
    content = file.read()
    print(f"w+ mode content: {content}")

#Demonstrates r+

with open("/content/rplus.txt", "w+") as file:
    file.write("This is r+ mode demonstration\n")
    file.seek(0)
    content = file.read()
    print(f"r+ mode content: {content}")



# Q5
binary_file_path = "/content/bin"

with open(binary_file_path, "wb") as file:
    file.write(b"These are some bytes.")

with open(binary_file_path, "ab") as file:
    file.write(b"Appended bytes.")

with open(binary_file_path, "rb") as file:
    contents = file.read()
    print(f"Binary file contents: {contents}")


# Q6
def rd(file_path):
    try:
        with open(file_path, "r") as file:
            content = file.read(4)  # Read the first four characters
            return content
    except FileNotFoundError:
        return "File not found."

# Get the file path from the user
user_file_path = input("Enter the file path: ")

# Call the rd function and print the result
result = rd(user_file_path)
print(f"First four characters: {result}")
```

```python
# Q1 Add the following data in a dataframe df using pandas.
# Name	Age	Sex
# Bharat Singh	21	M
# Priya Singh	18	F
# John Adams	19	M
# Nita Jain	20	F
# (a)	Print the rows 2-4 in df.
# (b)	If age is less than 20 then for all rows modify the age to 15 in the dataframe.
# (c)	Change the age of row 4 in df to 40.

import pandas as pd

# Q1
data = {'Name': ['Bharat Singh', 'Priya Singh', 'John Adams', 'Nita Jain'],
        'Age': [21, 18, 19, 20],
        'Sex': ['M', 'F', 'M', 'F']}
df = pd.DataFrame(data)

# (a) Print rows 2-4 (Note: Python uses 0-based indexing)
print("(a) Rows 2-4:")
print(df.iloc[1:4])  # Rows with index 1, 2, and 3

# (b) Modify age to 15 if age < 20
df.loc[df['Age'] < 20, 'Age'] = 15
print("\n(b) Modified DataFrame:")
print(df)

# (c) Change age of row 4 to 40 (index 3)
df.loc[3, 'Age'] = 40
print("\n(c) Age of row 4 changed:")
df
```

```python
# Q2 Perform the following using pandas, on a CSV file.
# (a)	Add the contents of this CSV file in a dataframe df.
# (b)	Remove the rows containing null values from this dataframe df.
# (c)	Remove all the duplicate rows from df.

import pandas as pd

# Assuming your CSV file is named 'your_file.csv' and is in the /content directory.
# Replace 'your_file.csv' with the actual file name and path if needed.
file_path = '/content/your_file.csv'

try:
  # (a) Add the contents of the CSV file to a DataFrame df.
  df = pd.read_csv(file_path)

  # (b) Remove rows with null values.
  df = df.dropna()

  # (c) Remove duplicate rows.
  df = df.drop_duplicates()

  # Print the modified DataFrame (optional).
  print(df)

except FileNotFoundError:
  print(f"Error: File '{file_path}' not found.")
except pd.errors.EmptyDataError:
  print(f"Error: File '{file_path}' is empty.")
except pd.errors.ParserError:
  print(f"Error: Could not parse the CSV file '{file_path}'. Check the file format.")
```
