```python
# Q Implement the following.
# •	Create an empty set S.
# •	Add some elements in this set  S i.e 1,2,1.5,7.
# •	Find the number of elements in this set.
# •	Remove 1 from S.
# •	Create a copy of set S.
# •	Create a set S2={1,9,9,0}
# •	Find set difference, union, intersection of S and S2.
# •	Check if 12 is present in set S.

# Q1
S = set()
S.add(1)
S.add(2)
S.add(1.5)
S.add(7)

print("Number of elements in S:", len(S))

S.remove(1)
S_copy = S.copy()
S2 = {1, 9, 9, 0}

set_difference = S - S2
set_union = S | S2
set_intersection = S & S2

print("Set difference:", set_difference)
print("Set union:", set_union)
print("Set intersection:", set_intersection)

print("Is 12 present in S?", 12 in S)

# Q2
A = {10, 20, 30, 40, 50}
B = {40, 50, 60, 70, 80}

union_AB = A | B
intersection_AB = A & B
difference_AB = A - B
symmetric_difference_AB = A ^ B

print("Union:", union_AB)
print("Intersection:", intersection_AB)
print("Difference (A - B):", difference_AB)
print("Symmetric Difference:", symmetric_difference_AB)


# Q3
C = {5, 10, 15}
D = {5, 10, 15, 20, 25}

print("Is C a subset of D?", C.issubset(D))
print("Is D a superset of C?", D.issuperset(C))


# Q4
E = {1, 2, 3, 4, 5}
E.add(6)
E.remove(3)
E.update({7, 8, 9})
print(E)


# Q5
numbers = [10, 20, 30, 20, 10, 40, 50, 60, 50, 60]
distinct_numbers = set(numbers)
print("Distinct elements:", distinct_numbers)


# Q6
X = {100, 200, 300, 400, 500}
Y = {300, 400, 500, 600, 700}
print("Elements in both sets:", X & Y)
print("Elements in X but not in Y:", X - Y)
print("Elements in Y but not in X:", Y - X)
print("Are the sets disjoint?", X.isdisjoint(Y))
```
